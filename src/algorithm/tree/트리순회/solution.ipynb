```python
nodes = test_cases[0].input['nodes']
node_list = [0] + nodes
```

```python
# 전위 순회. 본인 -> 왼쪽 -> 오른쪽

def pre_search(node_list):
    current_node = 1
    result = []
    visited = [False] * len(node_list)

    # 조건에 노드가 있는지 확인하는게 빠졌다.

    while True:
        # 본인 방문할 수 있으면 방문
        if not visited[current_node]:
            result.append(node_list[current_node])
            visited[current_node] = True
        # 본인 방문했으면 왼쪽 시도
        # 갈 수 있으면 이동
        elif current_node * 2 < len(node_list) and not visited[current_node * 2]:
            current_node = current_node * 2 
        elif current_node * 2 < len(node_list) and not visited[current_node * 2 + 1] :
            current_node = current_node * 2 + 1
        elif current_node // 2 > 0: # 모두 다 방문해서 갈 곳이 없으면 부모 노드로 올라가기
            current_node = current_node // 2
        else:
            break
        
    return result
```

```python
pre_search(node_list)
```

```python
# 중위 순회. 왼쪽 -> 본인 -> 오른쪽

def mid_search(node_list):
    current_node = 1
    result = []
    visited = [False] * len(node_list)

    # 조건에 노드가 있는지 확인하는게 빠졌다.

    while True:
        # 아 순서 차이구나.
        if current_node * 2 < len(node_list) and not visited[current_node * 2]:
            current_node = current_node * 2 
        # 본인 방문할 수 있으면 방문
        elif not visited[current_node]:
            result.append(node_list[current_node])
            visited[current_node] = True
        # 본인 방문했으면 왼쪽 시도
        # 갈 수 있으면 이동

        elif current_node * 2 < len(node_list) and not visited[current_node * 2 + 1] :
            current_node = current_node * 2 + 1
        elif current_node // 2 > 0: # 모두 다 방문해서 갈 곳이 없으면 부모 노드로 올라가기
            current_node = current_node // 2
        else:
            break
        
    return result
```
