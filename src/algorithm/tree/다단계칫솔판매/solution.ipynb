```python
# 트리 쪽 확실히 잘 안풀리네
# 이런 문제를 보면 이진트리를 그래서 구현해야하는건가? 라는 생각이 듬

enroll, referral, seller, amount = test_cases[0].input.values()
```

```python
enroll, referral, seller, amount
```

```python
# build tree with enroll, referral
# set으로 관리하는게 훨씬 편해보이는데, 자식을 담고있는

from collections import defaultdict

childs_dict = defaultdict(list)
seller_dict = defaultdict(int)

for en, re in zip(enroll, referral):
    childs_dict[re].append(en)
    
for se, am in zip(seller, amount):
    seller_dict[se] = am
```

```python
seller_dict['mk']
```

```python

def get_profit(name):
    # 본인의 판매금의 90% + 자식의 판매금의 10%
    childs_profit_sum = 0
    for child in childs_dict[name]:
        childs_profit_sum += int(get_profit(child) * 0.1)
    
    # 본인 이익 
    byself_profit = seller_dict[name] * 100 - int(seller_dict[name] * 100 * 0.1)
    
    return byself_profit + childs_profit_sum
```

```python
result = []

for se in enroll:
    result.append(get_profit(se))
```
