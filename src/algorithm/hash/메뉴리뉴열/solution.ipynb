```python
orders, course = example1.input.values()
orders, course
```

```python
from itertools import combinations
from collections import Counter

counter = Counter()

for order in orders:
    sorted_order = sorted(order)
    for size in course:
        for combo in combinations(sorted_order, size):
            key = "".join(combo)
            counter[key] += 1
```
