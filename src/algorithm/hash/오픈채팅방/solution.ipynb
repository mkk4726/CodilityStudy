```python
# 채팅방의 id : 닉네임 hash를 관리하는 알고리즘이 필요한거네?
# hash에 저장해두는 이유는 결국 최종 닉네임을 알려고
# 나갔다가 들어오면 새로운 닉네임을 부여해야 함 -> 이 처리를 어떻게 할 수 있으려나? 다시 들어오는건 무조건 나갔다는 이야기니까
# 나가는건 굳이 처리 안해도 됨
# 엥 바꾸는 것도 uid랑 매핑만 바꾸면 되는거 아닌가?

record_hash = {}

for rec in record:
    rec_list = rec.split(" ")
    action, uid = rec_list[0], rec_list[1]
    
    if action in ["Enter", "Change"]:
        record_hash[uid] = rec_list[2]
```

```python
def solution(record):
    record_hash = {}

    for rec in record:
        rec_list = rec.split(" ")
        action, uid = rec_list[0], rec_list[1]
        
        if action in ["Enter", "Change"]:
            record_hash[uid] = rec_list[2]
            
    expected = []

    for rec in record:
        rec_list = rec.split(" ")
        action, uid = rec_list[0], rec_list[1]
        if action == "Change":
            continue
        
        action_str = "들어왔습니다." if action == "Enter" else "나갔습니다."
        
        expected.append(f"{record_hash[uid]}님이 {action_str}")
    
    return expected
```

```python
results = run_testcases(solution, test_cases)
results
```
