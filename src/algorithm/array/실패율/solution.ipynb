```python
example1 = test_cases[0].input
```

```python
N, stages = example1['n'], example1['stages']
```

```python
fail_rate = []

for i in range(1, N+1):
    on_going_cnt = 0 # i == stage
    after_all_cnt = 0 # i >= stage

    for stage in stages:
        if stage == i:
            on_going_cnt += 1
        if stage >= i:
            after_all_cnt += 1

    fail_rate.append([i, on_going_cnt / after_all_cnt])

fail_rate
```

```python
[x[0] for x in sorted(fail_rate, key=lambda x: x[1], reverse=True)]
```

```python
# fail cnt per stage
fail_cnt_array = [0] * (N + 2)
for stage in stages:
    fail_cnt_array[stage] += 1
```

```python
fail_cnt_array
```

```python
total_cnt = sum(fail_cnt_array)
total_cnt
```

```python
fail_rate_array = []

for i in range(1, N + 1):
    fail_cnt = fail_cnt_array[i]
    if fail_cnt == 0:
        fail_rate = 0
    else:
        fail_rate = fail_cnt / total_cnt
        total_cnt -= fail_cnt

    fail_rate_array.append([i, fail_rate])
```
