```python
example1 = test_cases[0].input

arr1, arr2 = example1['arr1'], example1['arr2']
print(arr1)
print(arr2)
```

```python
# 3 x 2 @ 2 x2 = 3 x 2
# 2 x2 행렬을 만들고 하나씩 채울까?

m, n = len(arr1), len(arr2[0])

result = [[-1 for _ in range(n)] for _ in range(m)]
result
```

```python
for i in range(m):
    for j in range(n):
        tmp = 0
        # 행이랑 열을 곱하기
        for k in range(len(arr2)):
            tmp += arr1[i][k] * arr2[k][j]
        result[i][j] = tmp

result
```
