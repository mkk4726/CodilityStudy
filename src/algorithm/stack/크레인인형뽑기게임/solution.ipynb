```python
board_stacks = [[]]

for i in range(len(board)):
    board_stacks.append([column[i] for column in board if column[i] > 0][::-1])  # 0번째 열만 추출하려면
```

```python
board_stacks
```

```python
moves
```

```python
result_stack = []
delete_cnt = 0

for move in moves:
    print("move:", move)
    print("board_stack[move]", board_stacks[move])
    if len(board_stacks[move]) == 0:
        continue

    top = board_stacks[move].pop()
    if len(result_stack) > 0 and result_stack[-1] == top:
        result_stack.pop()
        delete_cnt += 1
    else:
        result_stack.append(top)
    print("result_stack:", result_stack)
    print("delete_cnt:", delete_cnt)
```
