```python
from pathlib import Path
from src.core.testcase import load_testcases, run_testcases
from solution import solution

test_cases = load_testcases(Path.cwd() / "testcases.txt")
results = run_testcases(solution, test_cases)
prices  = test_cases[0].input['prices']
```

```python
# 아직 할인 짝을 못 찾은 인덱스들 (오름차순 유지)

answer = prices.copy()
stack: list[int] = []

for i, price in enumerate(prices):
  print(i)
  print(stack)
  while stack and prices[stack[-1]] >= price:
      k = stack.pop()
      answer[k] -= price
  stack.append(i)
  print(stack)
```
