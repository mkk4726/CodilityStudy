```python
stack = []
answer = [0] * len(temperatures)

for i, temperature in enumerate(temperatures):
    print("i:", i)
    print(stack)
    while len(stack) > 0 and temperatures[stack[-1]] <= temperature:
        k = stack.pop()
        print("k:", k)
        answer[k] = i - k
    stack.append(i)
```
