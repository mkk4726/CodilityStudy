```python
i = 1
n, k, cmd = test_cases[i].input.values()
print(test_cases[i])
```

```python
# 현재 위치 관리
# D X , U X -> 이동하는거 구현
# C 삭제 후 위치
# 삭제헀던 위치 관리 -> X로 되돌렸을 떄 살릴 수 있음 (보고 있는 위치는 안바뀜 )


# 위치 이동을 어떻게 관리하지?
# index 를 가지고 있는 배열을 두고 거기서 꺼내야 하나?
# 밑으로 2칸 이동했을 때의 위치가 뭔지 어떻게 알지?
# 

deleted_index = []

curr_loc = k

def move_down(curr_loc, d_down):
    while d_down > 0 and curr_loc < n-1:
        curr_loc += 1
        if curr_loc not in deleted_index:
            d_down -= 1

    return curr_loc

def move_up(curr_loc, d_up):
    while d_up > 0 and curr_loc > 0:
        curr_loc -= 1
        if curr_loc not in deleted_index:
            d_up -= 1

    return curr_loc

for command in cmd:
    print("before loc:", curr_loc)
    print("cmd", command)
    if 'D' in command:
        curr_loc = move_down(curr_loc, int(command[-1]))
    elif 'U' in command:
        curr_loc = move_up(curr_loc, int(command[-1]))
    elif 'C' == command: # 여기서 틀렸네. 지우고나서 아래로 못가면 한칸 위로 가야하는데
        deleted_index.append(curr_loc)
        next_loc = move_down(curr_loc, 1)
        if curr_loc == next_loc:
            next_loc = move_up(curr_loc, 1)
        curr_loc = next_loc
    else: # Z 되돌리기
        deleted_index.pop()
    print("after loc: ", curr_loc)
    print("deleted_index: ", deleted_index)
    print("-------")
```
