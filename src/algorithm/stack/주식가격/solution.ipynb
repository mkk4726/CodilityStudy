```python
# 0번째의 초를 세려면
# N-1번째까지 비교를 해야함 (최대)

# 1번쨰 초도 마찬가지
# 그냥 for문으로 풀면 N (N-1)

# 이걸 O(N)으로 줄이려면

# stack 유형이라고 하니까 스택으로 생각해보게 되긴 하네 (힌트를 얻은거나 마찬가지)
```

```python
stack = []
len_prices = len(prices)

result = [0] * len_prices

for i, price in enumerate(prices):
    if len(stack) == 0:
        stack.append((i, price))
    else:
        while stack[-1][1] > price:
            past_i, past_price = stack[-1]
            result[past_i] = i - past_i
            stack.pop()
        stack.append((i, price))

while stack:
    top = stack.pop()
    i, price = top
    result[i] = len_prices - i - 1
```
