```python
from pathlib import Path

from src.core.testcase import load_testcases

test_cases = load_testcases(Path.cwd() / "testcases.txt")
```

```python
test_cases
```

```python
exampl1 = test_cases[0]

progresses, speeds = exampl1.input.values()
```

```python
# 각 작업별로 그래서 몇일이 걸리는지를 계산해야됨

days = []

for progress, speed, in zip(progresses, speeds):
    left_progress = 100 - progress
    day = left_progress / speed
    day = int(day) if int(day) == day else int(day) + 1
    days.append(day)
```

```python
days

# 앞 작업이 끝나야 다음 작업을 끝낼 수 있음
# 음 이거 어떻게 짜지? 감이 안잡히는데
```

```python
# 음! 큐에서 빼면서 뺸 값이랑 다음 값 비교하면서 채우면 되겠네

result = []

f_day = days.pop(0)
cnt = 1
    
while days:
    if days[0] <= f_day:
        cnt += 1
        days.pop(0)
    else:
        result.append(cnt)
        f_day = days.pop(0)
        cnt = 1

result.append(cnt)

result
```
