```python
from pathlib import Path

from src.core.testcase import load_testcases

test_cases = load_testcases(Path.cwd() / "testcases.txt")
example1 = test_cases[0]
cards1, cards2, goal = example1.input.values()
```

```python
# 만들 수 있냐? 


answer = "Yes"
while goal:
    target_word = goal.pop(0)
    print(target_word)
    if cards1[0] == target_word:
        cards1.pop(0)
    elif cards2[0] == target_word:
        cards2.pop(0)
    else:
        answer = "No"
        break
```
