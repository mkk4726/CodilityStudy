```python
example = test_cases[0]
n, k = example.input.values()
```

```python
# 기준으로 K번째 제거

from collections import deque

dq = deque(range(1, n+1))
lst = list(range(1, n+1))
```

```python
dq
```

```python
while len(dq) != 1:
    # 앞에서 K번째 뽑기
    # popleft(2) 이런건 없는거고
    # k만큼 뽑고 맨마지막 제외하고는 다시 넣어야하는데?
    for i in range(1, k+1):
        left = dq.popleft()
        if i != k:
            dq.append(left)
dq
```
